```python
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np


beta=1
def softmax(x):
    e_x = np.exp((x - np.max(x)) * beta)
    return e_x / e_x.sum()
#def softmax(x):
    #total = sum(x)
    #return [val ** beta / (total ** beta) for val in x]


def perpendicular_foot(Px, Py, Ox, Oy, m): # Foot of the perpendicular from point P to line with slope m and passing through O
    Px_, Py_ = sp.symbols('Px_ Py_')
    
    eq1 = sp.Eq(Py - Py_, (-1/m) * (Px - Px_))
    eq2 = sp.Eq(Oy - Py_, m * (Ox - Px_))
    
    solution = sp.solve((eq1, eq2), (Px_, Py_))
    
    return float(solution[Px_]), float(solution[Py_])

def signed_l(Ax,Ay,Bx,By):
    # Calculate the distance between the points
    l = np.sqrt((Ax - Bx)**2 + (Ay - By)**2)

    # Calculate the signed length based on the formula
    if (Ax - Bx) != 0:
        signed_l = l * (Ax - Bx) / np.abs(Ax - Bx)
    else:
        signed_l = 0
    return signed_l


Ox,Oy = 2,5
# Given points
Ax, Ay = 2, 5
Bx, By = 5, 2

D1x,D1y = 0.5*Ax, Ay # Range Decoy targeting A
D2x,D2y = Bx, 0.5*By # Range Decoy targeting B
D11x,D11y = 0.5*By+Bx,By

# Calculate Me (slope of AB)
Me = (Ay - By) / (Ax - Bx)
a_b_ratio_c1_softmax = []
a_b_ratio_c2_softmax = []
Mi = -0.6
Ox,Oy = 2,5
# List of c values for a fixed Mi
Oy_values = np.arange(Oy -5, Oy +5, 0.2)

def normalize(x):
    total = sum(x)
    return [val ** beta / (total ** beta) for val in x]
a_b_ratio_c1_normalized = []
a_b_ratio_c2_normalized = []


for Oy in Oy_values:
    Bx_, By_ = perpendicular_foot(Bx, By,Ox,Oy, Mi)
    Db = signed_l(Bx,By,Bx_,By_)

    Ax_, Ay_ = perpendicular_foot(Ax, Ay,Ox,Oy,Mi)
    Da = signed_l(Ax,Ay,Ax_,Ay_)

    D1x_,D1y_ = perpendicular_foot(D1x, D1y,Ox,Oy, Mi)
    d1 = signed_l(D1x,D1y,D1x_,D1y_)
    choice_c1 = softmax([Da, Db, d1])

    D2x_,D2y_ = perpendicular_foot(D2x, D2y,Ox,Oy, Mi)
    d2 = signed_l(D2x,D2y,D2x_,D2y_)
    choice_c2 = softmax([Da, Db, d2])

    a_b_ratio_c1 = choice_c1[0]/choice_c1[1]
    a_b_ratio_c2 = choice_c2[0]/choice_c2[1]
    a_b_ratio_c1_softmax.append(a_b_ratio_c1)
    a_b_ratio_c2_softmax.append(a_b_ratio_c2)
    

for Oy in Oy_values:
    Bx_, By_ = perpendicular_foot(Bx, By,Ox,Oy, Mi)
    Db = signed_l(Bx,By,Bx_,By_)

    Ax_, Ay_ = perpendicular_foot(Ax, Ay,Ox,Oy,Mi)
    Da = signed_l(Ax,Ay,Ax_,Ay_)

    D1x_,D1y_ = perpendicular_foot(D1x, D1y,Ox,Oy, Mi)
    d1 = signed_l(D1x,D1y,D1x_,D1y_)
    choice_c1 = normalize([Da, Db, d1])

    D2x_,D2y_ = perpendicular_foot(D2x, D2y,Ox,Oy, Mi)
    d2 = signed_l(D2x,D2y,D2x_,D2y_)
    choice_c2 = normalize([Da, Db, d2])

    a_b_ratio_c1 = choice_c1[0]/choice_c1[1]
    a_b_ratio_c2 = choice_c2[0]/choice_c2[1]
    a_b_ratio_c1_normalized.append(a_b_ratio_c1)
    a_b_ratio_c2_normalized.append(a_b_ratio_c2)
    
# Set up the plot
plt.figure(figsize=(8, 6))

plt.plot(Oy_values, a_b_ratio_c1_softmax, marker='o', color='red', label='P(A|C1)/P(B|C1) using Softmax')
plt.plot(Oy_values, a_b_ratio_c2_softmax, marker='x', color='blue', label='P(A|C2)/P(B|C2) using Softmax')

plt.xlabel('Y Intercept, C', fontsize=14, weight='bold')
plt.ylabel('P(A)/P(B)', fontsize=14, weight='bold')
# plt.title('Softmax as a model of Choice')
plt.plot(Oy_values, a_b_ratio_c1_normalized, marker='o', color='yellow', label='P(A|C1)/P(B|C1) using Normalization')
plt.plot(Oy_values, a_b_ratio_c2_normalized, marker='x', color='green', label='P(A|C2)/P(B|C2) using Normalization')


# Legend settings
plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.1),ncol = 2, fontsize=12)
plt.grid(True)
plt.show()
```
